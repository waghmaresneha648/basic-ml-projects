```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# 1. Dataset: Unlabeled Store Customer Data
# Features: Age, Annual Income ($k), Spending Score (1-100)
data = {
    'Age': [19, 21, 20, 23, 31, 22, 35, 23, 64, 30, 67, 35, 58, 24, 37],
    'AnnualIncome': [15, 15, 16, 17, 17, 18, 19, 19, 19, 19, 20, 20, 20, 21, 21],
    'SpendingScore': [39, 81, 6, 77, 40, 76, 6, 94, 3, 72, 14, 99, 15, 77, 5]
}

df = pd.DataFrame(data)

# 2. Scale Features
# Clustering relies heavily on distance calculations, so scaling is critical
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)

# 3. Train K-Means Model
# We ask K-Means to divide our customers into 3 distinct clusters (K=3)
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df['Cluster'] = kmeans.fit_predict(X_scaled)

print("Customer Data with Assigned Clusters:")
print(df)
print("\n" + "="*40 + "\n")

# 4. Analyze Cluster Means
cluster_summary = df.groupby('Cluster').mean()
print("Average Values per Cluster:")
print(cluster_summary)
print("\n" + "="*40 + "\n")

# 5. Predict Cluster for a New Customer
# FIX: Wrap input in a DataFrame with matching column names (prevents UserWarning)
new_customer_df = pd.DataFrame(
    [[25, 18, 80]], 
    columns=['Age', 'AnnualIncome', 'SpendingScore']
)

# Scale and predict
new_customer_scaled = scaler.transform(new_customer_df)
assigned_cluster = kmeans.predict(new_customer_scaled)[0]

print(f"New Customer assigned to Cluster: {assigned_cluster}")
```
