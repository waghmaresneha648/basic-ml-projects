```python
import random
import os

SCORE_FILE = "highscore.txt"

def load_high_score():
    if os.path.exists(SCORE_FILE):
        with open(SCORE_FILE, "r") as f:
            try:
                return int(f.read().strip())
            except ValueError:
                return float("inf")
    return float("inf")

def save_high_score(score):
    with open(SCORE_FILE, "w") as f:
        f.write(str(score))

def play_game():
    secret_number = random.randint(1, 100)
    attempts = 0
    best_score = load_high_score()

    print("=== Number Guessing Game ===")
    print("I've picked a number between 1 and 100.")
    if best_score != float("inf"):
        print(f"Current High Score (fewest attempts): {best_score}")
    else:
        print("No high score yet. Set the first record!")
    print("-" * 35)

    while True:
        try:
            guess = int(input("Enter your guess (1-100): "))
        except ValueError:
            print("Please enter a valid integer.")
            continue

        attempts += 1

        if guess < secret_number:
            print("Too low! Try higher.")
        elif guess > secret_number:
            print("Too high! Try lower.")
        else:
            print(f"\nCorrect! You guessed it in {attempts} attempts.")
            if attempts < best_score:
                print("New High Score achieved!")
                save_high_score(attempts)
            break

if __name__ == "__main__":
    play_game()
```
